# 01 · Data audit and analytical sample

Audit the revised sample without modifying original records. Q087 was excluded post hoc; corrected reruns are not part of the primary analysis.

```python
from pathlib import Path
import pandas as pd

# Run from the repository root or its notebooks/ directory.
cwd = Path.cwd().resolve()
ROOT = cwd.parent if cwd.name == "notebooks" else cwd
candidates = sorted(ROOT.rglob("retained_human_annotations_990.csv"))
if len(candidates) != 1:
    raise FileNotFoundError(
        f"Expected one revised analytical dataset, found {len(candidates)}. "
        "Extract the revised 99-item package into the repository and remove duplicate copies."
    )
DATA = candidates[0].parent
print("Repository:", ROOT)
print("Analysis directory:", DATA)

def load(name):
    return pd.read_csv(DATA / name, encoding="utf-8-sig")
```

```python
human = load("retained_human_annotations_990.csv")
executions = load("retained_executions_1188.csv")
assert len(human) == 990
assert len(executions) == 1188
for frame in (human, executions):
    assert frame["item_id"].nunique() == 99
    assert not frame["item_id"].astype(str).str.upper().eq("Q087").any()
display(human.groupby("condition").size().rename("human_rated_n").to_frame())
assert human.groupby("condition").size().to_dict() == {"C1":99,"C2":297,"C3":297,"C4":297}
```

```python
excluded_h = load("excluded_Q087_annotations_10.csv")
excluded_e = load("excluded_Q087_executions_12.csv")
assert len(excluded_h) == 10 and len(excluded_e) == 12
display(pd.DataFrame({"sample":["retained human","retained executions","excluded human","excluded executions"],"n":[len(human),len(executions),len(excluded_h),len(excluded_e)]}))
```

Repeated baseline execution records are not independent observations. This audit checks the retained sample; it does not perform new annotation or inference.
