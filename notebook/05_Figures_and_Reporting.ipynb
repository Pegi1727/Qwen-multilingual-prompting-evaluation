# 05 · Figures and reporting

Display the existing repository figures without renaming or renumbering them. Images are expected in figures/, not in the repository root.

```python
from pathlib import Path
from IPython.display import display, Image, Markdown
cwd = Path.cwd().resolve()
ROOT = cwd.parent if cwd.name == "notebooks" else cwd
FIGURES = ROOT / "figures"
files = [
    ("Figure 1 · Metric comparison", "Figure_1_Metric_Comparison.png"),
    ("Figure 2 · Content distribution", "Figure_2_Content_Distribution.png"),
    ("Figure 3 · Exact match and confidence intervals", "Figure_3_Exact_Match_Confidence_Intervals.png"),
    ("Figure 4 · Language-specific outcomes", "Figure_4_Language_Specific_Outcomes.png"),
    ("Figure 5 · Content by format", "Figure_5_Content_by_Format.png"),
    ("Graphical abstract", "Qwen_Graphical_Abstract_99_Items.png")
]
for title, name in files:
    path = FIGURES / name
    if not path.is_file():
        raise FileNotFoundError(f"Missing expected image: {path}")
    display(Markdown("## " + title))
    display(Image(filename=str(path),width=1000))
```
