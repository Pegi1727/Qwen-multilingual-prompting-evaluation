# 02 · Human-rated outcomes

Inspect the archived human-evaluation summaries. Human-rated results are descriptive and originate from one evaluator.

```python
from pathlib import Path
import pandas as pd

# Run from the repository root or its notebooks/ directory.
cwd = Path.cwd().resolve()
ROOT = cwd.parent if cwd.name == "notebooks" else cwd
candidates = sorted(ROOT.rglob("retained_human_annotations_990.csv"))
if len(candidates) != 1:
    raise FileNotFoundError(
        f"Expected one revised analytical dataset, found {len(candidates)}. "
        "Extract the revised 99-item package into the repository and remove duplicate copies."
    )
DATA = candidates[0].parent
print("Repository:", ROOT)
print("Analysis directory:", DATA)

def load(name):
    return pd.read_csv(DATA / name, encoding="utf-8-sig")
```

```python
by_condition = load("qwen_final_by_condition.csv")
display(by_condition)
display(load("qwen_final_by_condition_language.csv"))
display(load("qwen_content_by_format.csv"))
assert by_condition["correct_n"].sum() == 321
assert by_condition["format_compliant_n"].sum() == 362
assert by_condition["correct_and_compliant_n"].sum() == 130
```

```python
import matplotlib.pyplot as plt
metrics = ["correct_pct", "format_compliant_pct", "correct_and_compliant_pct"]
ax = by_condition.set_index("condition")[metrics].rename(columns={
    "correct_pct":"Content correctness", "format_compliant_pct":"Format compliance",
    "correct_and_compliant_pct":"Joint success"
}).plot.bar(figsize=(10,5), rot=0)
ax.set(ylabel="Responses (%)", xlabel="Condition", ylim=(0,100), title="Human-rated outcomes · revised 99-item analysis")
plt.tight_layout()
plt.show()
```

Mixed responses are not counted as correct. Content correctness, format compliance, and joint success measure different outcomes; no inter-rater agreement estimate is available.
